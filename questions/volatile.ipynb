## How are read-only and write-only accesses enforced? (enforcing)
The `ReadVolatile` and `WriteVolatile` types make it impossible to write and read, respectively, the underlying pointer. 

How do they accomplish this?

#### Answer

The `ReadVolatile` and `WriteVolatile` types enforce read-only and write-only access by restricting how their inner pointer (`self.0`) is exposed. Since this pointer is private, external code cannot directly access it. Instead, `ReadVolatile` provides a `read` method and only exposes a `*const T`, preventing modifications, while `WriteVolatile` provides a `write` method and returns a `*mut T`, allowing writes but potentially enabling indirect reads. This design ensures controlled access while still allowing volatile memory operations.

## What do the macros do? (macros)
What do the `readable!`, `writeable!`, and `readable_writeable!` macros do?

#### Answer

The `readable!`, `writeable!`, and `readable_writeable!` macros are helper macros that simplify the implementation of traits for `Volatile` types. They automatically generate the necessary trait implementations for `ReadVolatile`, `WriteVolatile`, or both, reducing boilerplate code. By using these macros, the codebase remains cleaner and more maintainable while ensuring consistent behavior across different `Volatile` types.
